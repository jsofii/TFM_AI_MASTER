```python
import torch
import torch.nn as nn
import librosa
import numpy as np
import torch.nn.functional as F

# Definir la clase del modelo LSTM
class EmotionRecognitionLSTM(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, num_classes):
        super(EmotionRecognitionLSTM, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # Definir la capa LSTM
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        
        # Definir la capa fully connected (FC)
        self.fc = nn.Linear(hidden_size, num_classes)
    
    def forward(self, x):
        # Inicializar los estados ocultos y las celdas de memoria para la LSTM
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)
        
        # Pasar los datos a través de la capa LSTM
        out, _ = self.lstm(x, (h0, c0))
        
        # Tomar la salida de la última celda LSTM y pasarla por la capa fully connected
        out = self.fc(out[:, -1, :])
        
        return out

# Cargar el modelo guardado
input_size = 38  # Ajustamos a 38 características por timestep
hidden_size = 256
num_layers = 3
num_classes = 8  # Número de emociones

# Instanciar el modelo
model = EmotionRecognitionLSTM(input_size=input_size, hidden_size=hidden_size, num_layers=num_layers, num_classes=num_classes)

# Cargar los pesos entrenados
model.load_state_dict(torch.load("final_model.pth"))

# Colocar el modelo en modo evaluación
model.eval()

# Función para extraer características de un archivo de audio con 38 características por timestep
def extract_features_from_file(audio_file, n_timesteps=100):
    # Cargar archivo de audio
    audio_data, sr = librosa.load(audio_file, sr=None)
    
    if len(audio_data) == 0:
        return np.array([])

    hop_length = len(audio_data) // n_timesteps
    
    # Extraer características con un total de 38 características
    mfccs = librosa.feature.mfcc(y=audio_data, sr=sr, n_mfcc=13, hop_length=hop_length)  # 13 MFCC
    chroma = librosa.feature.chroma_stft(y=audio_data, sr=sr, hop_length=hop_length)     # 12 Chroma
    spec_contrast = librosa.feature.spectral_contrast(y=audio_data, sr=sr, hop_length=hop_length)  # 7 Spectral contrast
    tonnetz = librosa.feature.tonnetz(y=librosa.effects.harmonic(audio_data), sr=sr, hop_length=hop_length)  # 6 Tonnetz
    
    # Combinar características para cada timestep (13 + 12 + 7 + 6 = 38 características)
    features = np.hstack([mfccs.T, chroma.T, spec_contrast.T, tonnetz.T])

    return features

# Función para predecir emociones a partir de un archivo de audio
def predict_emotion_from_file(audio_file, model):
    # Extraer características del archivo de audio
    features = extract_features_from_file(audio_file, n_timesteps=100)
    
    if features.size == 0:
        print("Error al extraer características del archivo de audio.")
        return None

    # Redimensionar para coincidir con la entrada del modelo
    features = torch.tensor(features).float().unsqueeze(0)

    # Predicción
    with torch.no_grad():
        outputs = model(features)
        probabilities = F.softmax(outputs, dim=1)
        predicted_label = torch.argmax(probabilities, dim=1).item()

    return predicted_label

# Ruta al archivo de audio WAV
audio_file = './testData/disgusto_sofi.wav'

# Predecir la emoción del archivo de audio
predicted_emotion = predict_emotion_from_file(audio_file, model)

if predicted_emotion is not None:
    print(f"Emoción predicha: {predicted_emotion}")
else:
    print("No se pudo procesar el archivo de audio.")
```

```python
import torch
import torch.nn as nn
import librosa
import numpy as np
import torch.nn.functional as F
import time
import pygame

# Diccionario para mapear la predicción numérica a la emoción
emotion_labels = {
    0: 'Neutralidad',
    1: 'Calma',
    2: 'Felicidad',
    3: 'Tristeza',
    4: 'Enojo',
    5: 'Miedo',
    6: 'Disgusto',
    7: 'Sorpresa'
}

# Definir la clase del modelo LSTM
class EmotionRecognitionLSTM(nn.Module):
    def __init__(self, input_size, hidden_size, num_layers, num_classes):
        super(EmotionRecognitionLSTM, self).__init__()
        self.hidden_size = hidden_size
        self.num_layers = num_layers

        # Definir la capa LSTM
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True)
        
        # Definir la capa fully connected (FC)
        self.fc = nn.Linear(hidden_size, num_classes)
    
    def forward(self, x):
        # Inicializar los estados ocultos y las celdas de memoria para la LSTM
        h0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)
        c0 = torch.zeros(self.num_layers, x.size(0), self.hidden_size).to(x.device)
        
        # Pasar los datos a través de la capa LSTM
        out, _ = self.lstm(x, (h0, c0))
        
        # Tomar la salida de la última celda LSTM y pasarla por la capa fully connected
        out = self.fc(out[:, -1, :])
        
        return out

# Cargar el modelo guardado
input_size = 38  # Ajustamos a 38 características por timestep
hidden_size = 256
num_layers = 3
num_classes = 8  # Número de emociones

# Instanciar el modelo
model = EmotionRecognitionLSTM(input_size=input_size, hidden_size=hidden_size, num_layers=num_layers, num_classes=num_classes)

# Cargar los pesos entrenados
model.load_state_dict(torch.load("final_model.pth"))

# Colocar el modelo en modo evaluación
model.eval()

# Función para extraer características de un fragmento de audio
def extract_features_from_audio_chunk(audio_chunk, sr, n_timesteps=100):
    if len(audio_chunk) == 0:
        return np.array([])

    hop_length = len(audio_chunk) // n_timesteps

    # Extraer características con un total de 38 características
    mfccs = librosa.feature.mfcc(y=audio_chunk, sr=sr, n_mfcc=13, hop_length=hop_length)  # 13 MFCC
    chroma = librosa.feature.chroma_stft(y=audio_chunk, sr=sr, hop_length=hop_length)     # 12 Chroma
    spec_contrast = librosa.feature.spectral_contrast(y=audio_chunk, sr=sr, hop_length=hop_length)  # 7 Spectral contrast
    tonnetz = librosa.feature.tonnetz(y=librosa.effects.harmonic(audio_chunk), sr=sr, hop_length=hop_length)  # 6 Tonnetz

    # Combinar características para cada timestep (13 + 12 + 7 + 6 = 38 características)
    features = np.hstack([mfccs.T, chroma.T, spec_contrast.T, tonnetz.T])

    return features

# Función para predecir emociones para un fragmento de audio
def predict_emotion_from_chunk(audio_chunk, sr, model):
    # Extraer características
    features = extract_features_from_audio_chunk(audio_chunk, sr)
    
    if features.size == 0:
        return None

    # Redimensionar para coincidir con la entrada del modelo
    features = torch.tensor(features).float().unsqueeze(0)

    # Predicción
    with torch.no_grad():
        outputs = model(features)
        probabilities = F.softmax(outputs, dim=1)
        predicted_label = torch.argmax(probabilities, dim=1).item()

    # Devolver el nombre de la emoción en lugar del número
    return emotion_labels.get(predicted_label, 'Emoción desconocida')

# Simular predicción en tiempo real mientras se reproduce el archivo de audio
def simulate_real_time_prediction(audio_file, model, chunk_duration=5.0):
    # Cargar el archivo de audio
    audio_data, sr = librosa.load(audio_file, sr=None)

    # Inicializar pygame para reproducir el audio
    pygame.mixer.init()
    pygame.mixer.music.load(audio_file)
    
    # Reproducir el archivo de audio
    pygame.mixer.music.play()

    # Dividir el audio en fragmentos
    chunk_samples = int(sr * chunk_duration)  # Convertir la duración del fragmento a muestras
    
    start = 0
    while start < len(audio_data):
        # Tomar un fragmento de audio
        end = min(start + chunk_samples, len(audio_data))
        audio_chunk = audio_data[start:end]

        # Predecir la emoción del fragmento
        predicted_emotion = predict_emotion_from_chunk(audio_chunk, sr, model)
        
        # Mostrar la predicción de la emoción (nombre en lugar de número)
        print(f"Emoción predicha: {predicted_emotion}")
        
        # Esperar el tiempo equivalente a la duración del fragmento
        time.sleep(chunk_duration)
        
        # Moverse al siguiente fragmento
        start = end

    # Esperar a que termine la reproducción del audio
    while pygame.mixer.music.get_busy():
        time.sleep(0.1)

    pygame.mixer.quit()

# Ruta al archivo de audio
audio_file = 'testData/test_sofi_full.wav'

# Simular predicción en tiempo real
simulate_real_time_prediction(audio_file, model, chunk_duration=1.0)
```
